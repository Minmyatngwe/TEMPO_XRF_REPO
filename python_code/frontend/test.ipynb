```python
import numpy as np

a1 = np.array([4, -2])
a2 = np.array([-3, 2])
A = np.array([a1, a2]).T

b1 = np.array([2, 4])
b2 = np.array([5, -2])
B = np.array([b1, b2]).T

v = np.array([5, 5])

# Coordinates of v in basis A
v_A = np.linalg.solve(A, v)

# B basis vectors expressed in A coordinates
c1 = np.linalg.solve(A, b1)
c2 = np.linalg.solve(A, b2)

C = np.array([c1, c2]).T

# Convert A coordinates into B coordinates
v_B = np.linalg.solve(C, v_A)

print("v in standard coordinates:", v)
print("v in A coordinates:", v_A)
print("v in B coordinates:", v_B)
print("Reconstructed standard vector:", B @ v_B)
```
